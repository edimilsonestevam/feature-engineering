<b>4. Código Fonte Python</b>

```python
# 1. Importando as bibliotecas
import pandas as pd
import numpy as np
```

```python
# 2. Criando um DataFrame
dados = {
    'idade': [25, 30, np.nan, 45, 35, np.nan],
    'salario': [2500, 3200, 2800, np.nan, 4500, 3900],
    'experiencia': [2, 5, 3, 10, np.nan, 7]
}
```

```python
# 3. Carregando os dados no DataFrame
df = pd.DataFrame(dados)
```

```python
# 4. Identificando os valores ausentes antes do tratamento
print("\nValores ausentes antes do tratamento:")
print(df.isnull().sum())
```

```python
# 5. Visualizando o DataFrame antes do tratamento
print("\nDataFrame antes do tratamento:")
df
```

```python
# 6. Calculando a mediana de cada variável
mediana_idade = df['idade'].median()
mediana_salario = df['salario'].median()
mediana_experiencia = df['experiencia'].median()
```

```python
# 7. Substituindo os valores ausentes pela mediana
df['idade'] = df['idade'].fillna(mediana_idade)
df['salario'] = df['salario'].fillna(mediana_salario)
df['experiencia'] = df['experiencia'].fillna(mediana_experiencia)
```

```python
# 8. Validando o resultado
print("\nDataFrame após o tratamento:")
print(df)
```

```python
# 9. Identificando os valores ausentes após o tratamento
print("\nValores ausentes após o tratamento:")
print(df.isnull().sum())
```

<b>8. Considerações sobre Aprendizado de Máquina</b>

<b>Objetivo:</b> demonstrar os cuidados necessários ao utilizar a imputação pela mediana durante o pré-processamento de dados para modelos de Aprendizado de Máquina.

1. Prevenção de Vazamento de Dados (Data Leakage)

A mediana deve ser calculada exclusivamente com os dados de treinamento (training set). Posteriormente, o mesmo valor deve ser aplicado aos conjuntos de validação (validation set) e teste (test set).

2. Separação dos Conjuntos de Dados

O fluxo recomendado:

3. Implementação com Scikit-learn

```python
# 1. Importando as bibliotecas
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split

# 2. Separando as variáveis
X = df[['idade', 'salario', 'experiencia']]

# 3. Dividindo os dados
X_train, X_test = train_test_split(
    X,
    test_size=0.2,
    random_state=42
)

# 4. Criando o pipeline de imputação
pipeline = Pipeline([
    ('imputacao', SimpleImputer(strategy='median'))
])

# 5. Calculando as medianas no treinamento
X_train_tratado = pipeline.fit_transform(X_train)

# 6. Aplicando as mesmas medianas no teste
X_test_tratado = pipeline.transform(X_test)
```
